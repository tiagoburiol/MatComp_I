```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
plt.rcParams["font.family"] = "serif"
plt.rcParams["mathtext.fontset"] = "dejavuserif"
```

```python
x = np.array([4, 5])
fig, ax = plt.subplots()

f = lambda x: 0.02+(x-1.0)**2/100
df = lambda x:(x-1)/50
xplot = np.linspace(x[0]-3,x[-1]+2)
ax.plot(xplot, f(xplot), '-',color='black')

# desenha uma reta tangente
x1 = 2.5; y1 = f(x[0])+df(x[0])*(x1-x[0])
x2 = 7.3; y2 = f(x[0])+df(x[0])*(x2-x[0])
ax.plot((x1,x2), (y1,y2), color='black', lw=0.5)


# desenha uma reta secante
m = (f(x[1])-f(x[0]))/(x[1]-x[0])

x1 = 2.7; y1 = f(x[0])+m*(x1-x[0])
x2 = 7.1; y2 = f(x[0])+m*(x2-x[0])
ax.plot((x1,x2), (y1,y2), color='black', lw=0.7)

# desenha as linhas verticais
ax.vlines(x=x[0],ymin=0,ymax=f(x[0]),
          color='black',linestyle='dotted', lw=1.2)
ax.plot(x[0], f(x[0])+0.0015, '.',color='black')
ax.vlines(x=x[1],ymin=0,ymax=f(x[1])+0.001,
          color='black',linestyle='dotted', lw=1.2)
ax.plot(x[1], f(x[1])+0.001, '.',color='black')

#texto e pontos no eixo x (ticks)
ax.text(x[0]-0.2 ,-0.04, "$p \longleftarrow$", style='italic', size='xx-large')
ax.text(x[1]-0.1 ,-0.04, "$x$", style='italic', size='xx-large')
ax.plot(x, np.zeros(len(x)), '.',color='black')

# desenha as linhas horizontais
ax.hlines(y=f(x[0]),xmin=0,xmax=x[0],
          color='black',linestyle='dotted', lw=1.2)
ax.plot(x[0], f(x[0])+0.0015, '.',color='black')
ax.hlines(y=f(x[1]),xmin=0,xmax=x[1],
          color='black',linestyle='dotted', lw=1.2)
ax.plot(x[1], f(x[1])+0.0015, '.',color='black')

#texto e pontos no eixo y (ticks)
ax.text(-1.2 ,f(x[0])-0.01, "$f(p)$", style='italic', size='xx-large')
ax.text(-1.2 ,f(x[1])-0.01, "$f(x)$", style='italic', size='xx-large')
ax.plot(np.zeros(len(x)), f(x), '.',color='black')


# Customizações para a figura final:
#____________________
# Limites em x e em y
ymin = -0.1; ymax = 0.4
xmin = -1; xmax = 8
plt.ylim(ymin, ymax) 
plt.xlim(xmin, xmax)
#____________________________________
# Remove as bordas
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)
#_______________________________________
# Desenha os eixos, setas e lable
ax.axhline(linewidth=1.0, color='black')
ax.plot((xmax-0.15), (0), marker=">", color='black')
ax.text(xmax-0.5,-0.04, "x", style='italic', size='xx-large')
ax.axvline(linewidth=1.0, color='black')
ax.plot((0), (ymax*0.99), marker="^",color='black')
ax.text(-0.5, ymax-0.04, "y", style='italic', size='xx-large')
      
#________________
# remove a escala
plt.xticks([])
plt.yticks([])
#plt.grid()
#_____________________________
# Salva em um arquivo
plt.savefig('png/derivada.png',dpi=200)
```

```python
def custonize(xmin=-1, xmax=13,
              ymin=-0.5, ymax=0.5,
              hideOY = True,
              filename='out.png'):
    
    
    # Customizações para a figura final:
    #____________________
    # Limites em x e em y
    #ymin = -0.5; ymax = 0.5
    #xmin = -1; xmax = 13
    plt.ylim(ymin, ymax) 
    plt.xlim(xmin, xmax)
    #____________________________________
    # Remove as bordas
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    #_______________________________________
    # Desenha os eixos, setas e lable
    ax.axhline(linewidth=1.0, color='black')
    ax.plot((xmax-0.15), (0), marker=">", color='black')
    ax.text(xmax-0.5,-ymax*0.1, "x", style='italic', size='large')
    
    if hideOY == False:
        ax.axvline(linewidth=1.0, color='black')
        ax.plot((0), (ymax*0.99), marker="^",color='black')
        ax.text(-0.2, ymax, "y", style='italic')
        
    #________________
    # remove a escala
    plt.xticks([])
    plt.yticks([])
    #plt.grid()
    #_____________________________
    # Salva em um arquivo
    plt.savefig(filename,dpi=200)
```

```python
plt.rcParams["font.family"] = "serif"
plt.rcParams["mathtext.fontset"] = "dejavuserif"
#['dejavusans', 'dejavuserif', 'cm', 'stix', 'stixsans', 'custom']

xplot = np.linspace(0.3, 5.6)
f = lambda x: np.exp(x/2)-1.56

x0 = 5.5
x1 = 4.3

xi = [x0,x1]
yi = [f(x0),f(x1)]

# plotando a função e a raiz
fig, ax = plt.subplots()

ax.plot(xplot, f(xplot),color='black')
ax.plot(0.85, 0, '.', color='black')



ax.text(xi[-2]-0.1, -0.7, '$x_0$', style='italic')
ax.text(xi[-1]-0.1, -0.7, '$x_1$', style='italic')
ax.text(0.76, -1.0, '$\overline{x}$', style='italic')


# Calcula as iterações
for i in range(1,5):
    # Calcula xi
    x = (x0*f(x1)-x1*f(x0))/(f(x1)-f(x0))
    print (x)
    
    # Alimenta a lista de pontos
    xi.append(x)
    yi.append(f(x))
    
    # Atualiza X0 e x1
    x0 = x1
    x1 = x
    
    # Desenha as retas secantes e os xi's
    ax.text(xi[-1]-0.1, -0.7, '$x_'+str(i+1)+'$', style='italic')
    ax.plot([xi[-3],xi[-1]], [f(xi[-3]),0], color='black', lw=0.5)
    
    # desenha as linhas verticais
    ax.vlines(x=xi, ymin=0, ymax=yi, 
          color='black', linestyle='dotted', lw=1.2)

                
# Customizações para a figura final:
#____________________
# Limites em x e em y
ymin = -2; ymax = 17
xmin = -0.5; xmax = 6
plt.ylim(ymin, ymax) 
plt.xlim(xmin, xmax)
#_____________________________________________________
# Inclui setas nos eixos nas posições certas
ax.plot((xmax-0.07), (0), marker=">", color='black')
ax.text(xmax-0.15,-1, "x", style='italic')
ax.plot((0), (ymax-0.3), marker="^",color='black')
ax.text(-0.2, ymax-0.7, "y", style='italic')
#____________________________________
# Remove as bordas
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)
#_______________________________________
# Desenha os eixos
ax.axhline(linewidth=1.0, color='black')
ax.axvline(linewidth=1.0, color='black')
#________________
# remove a escala
plt.xticks([])
plt.yticks([])
#plt.grid()
#_____________________________
# Salva em um arquivo
plt.savefig('png/secantes.png',dpi=300)
```
