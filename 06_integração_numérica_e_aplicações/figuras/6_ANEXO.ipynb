```python
def customizafig(nome="out.png",
                 xmin = -1, xmax = 7,
                 ymin = -0.1, ymax = 0.4,
                 dx = -0.5, dy=-0.04):

    # Customizações para a figura final:
    #____________________
    # Limites em x e em y
    #ymin = -0.1; ymax = 0.4
    #xmin = -1; xmax = 7
    plt.ylim(ymin, ymax) 
    plt.xlim(xmin, xmax)
    #____________________________________
    # Remove as bordas
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)
    #_______________________________________
    # Desenha os eixos, setas e lable
    ax.axhline(linewidth=1.0, color='black')
    ax.plot((xmax-0.1), (0), marker=">", color='black')
    ax.text(xmax+dx,dy, "x", style='italic', size='x-large')
    ax.axvline(linewidth=1.0, color='black')
    ax.plot((0), (ymax-0.007), marker="^",color='black')
    ax.text(dx, ymax+dy, "y", style='italic', size='x-large')
    
    #________________
    # remove a escala
    plt.xticks([])
    plt.yticks([])
    #plt.grid()
    #_____________________________
    # Salva em um arquivo
    plt.savefig(nome,dpi=200)
```
