## 2. Diferença entre as abordagens analítica e numérica

**Objetivo da aula:** Entender a diferença das abordagens analítica e numérica. 

As abordagens analítica e numérica, são duas maneiras diferentes de resolver problemas ou estudar fenômenos a partir de um modelo matemático.

**Abordagem analítica:** buscam-se soluções exatas, utilizando técnicas simbólicas e manipulações algébricas. Uma solução analítica, em geral, permite um maior entendimento conceitual sobre o problema ou fenômeno em estudo.

**Abordagem numérica:** usam-se métodos computacionais e algoritmos para obter soluções aproximadas para problemas onde a solução exata (analítica) é difícil ou impossível de encontrar. Em geral, transforma-se o problema contínuo em uma versão discretizada que é resolvida computacionalmente ou manualmente, com uma sequência de operações aritméticas relativamente simples.

A abordagem analítica costuma ser a primeira opção quando o problema é relativamente simples. Na prática, em problemas reais de engenharia, física, economia, etc, os procedimentos numéricos são mais utilizados. Em muitos projetos e pesquisas atuais, ambas as abordagens são usadas de forma complementar.

Em Python podemos podemos realizar manipulações algébricas por meio de [computação simbólica](https://docs.sympy.org/latest/tutorials/intro-tutorial/intro.html#what-is-symbolic-computation) usando a biblioteca [Sympy](https://www.sympy.org/). A computação simbólica trata da computação de objetos matemáticos simbolicamente. Isso significa que os objetos matemáticos são representados exatamente, não aproximadamente, e expressões matemáticas com variáveis ​​não avaliadas são mantidas em forma simbólica.

**Exemplo 8:**   Seja uma esfera de madeira de raio $r$ colocada a flutuar sobre a água conforme figura abaixo. Qual é a altura $d$ da esfera que ficará submersa? 


<img src="https://raw.githubusercontent.com/tiagoburiol/MatComp_I/refs/heads/main/02_abordagem_analitica_e_abordagem_numerica/figuras/esfera.png" width="200">

Para responder essa pergunta, observemos que, pelo *Princípio de Arquimedes* o peso da esfera deve ser equilibrado com o peso do volume de água deslocada, ou seja, $\mu_e V_e g = \mu_a V_a g$ onde $g$ é a aceleração da gravidade, $\mu_e$ é a densidade relativa da madeira e $\mu_a$ é a densidade relativa da água, $V_e$ e $V_a$ são os volumes. Essa equação nos leva a 

$$ \mu_e \frac{4 \pi r^3}{3} = \mu_a \int_0^d \pi \left[r^2-(x-r)^2 \right]dx $$

Em que o volume de água deslocada é calculado por meio de uma integral. Podemos obter uma expressão para o resultado da integral fazendo:

```python
d,x,mua,mue,r = sp.symbols('d, x, mu_a, mu_e,r',nonzero=True)

integ = mua*sp.integrate(sp.pi*(r**2-(x-r)**2), (x,0,d))
display(integ)
```

Então, escrever a equação:

```python
eq = sp.Eq(mue* 4*sp.pi*r**3/3, integ)
display(eq)
```

substituindo os valores $r=10 cm$,  $\mu_e =0,638 g/cm^3$ e $\mu_a=1 g/cm^3$, com $\pi \approx 3.1415$, podemos chegar à seguinte equação aproximada

$$ 1.047d^3-31.415d^2+2672.369=0$$

Observe aqui que usamos uma aproximação numérica para $\pi$. Basta resolver essa equação para obter a resposta procurada. 

No entanto, não existe uma fórmula simples para resolver essa equação. Além disso, observe que poderíamos ter multiplicado ambos os lados da equação por $\frac{3}{\pi}$, o que nos daria  
$$d^3-30d^2+2552=0$$

Com SymPy podemos fazer:

```python
dict = {r:10, mue:0.638, mua:1}
eq2 = sp.Eq(3 *eq.lhs.subs(dict), 3 *eq.rhs.subs(dict)) 
eq2 = sp.simplify(eq2)
eq2
```

que é uma equação mais precisa, uma vez que não tivemos que usar uma aproximação para $pi$, no entanto esse tipo de simplificaçao requer manipular o número $\pi$ como um símbolo.

No entanto não existe uma fórmula simples para resolver essa equação. Nesses casos recorremos aos métodos numéricos. Nesse caso específico, podemos encontrar a solução próxima a $d=11cm$ com a fórmula iterativa de ponto fixo:

$$d_{i+1} = \sqrt{\frac{-2552}{d_i-30}}$$
como é mostrado abaixo. Com 10 iterações obtemos a solução aproximada $d \approx 11,8615$ cm.

```python
d = 11
for i in range(10):
    d = math.sqrt(-2552/(d-30))
    print (d)
```

Por outro lado, poderíamos usar integração numérica para encontrar uma valor de $d$ que satisfaça a equação 

$$
\int_0^d \pi\left[r^2-(x-r)^2\right] dx = \frac{\mu_e}{\mu_a} \frac{4 \pi r^3}{3} = 0.638 \left(\frac{4\pi 1000}{3} \right) \approx 2672.448150653717
$$

testando para alguns valores de $d$. Um exemplo é mostrado a seguir:

```python
from scipy.integrate import quad

def f(x):
    return math.pi*(10**2-(x-10)**2)

for inc in range(10):
    d = 11.8 + 0.01*inc
    integral = quad(f,0,d)[0]
    erro = abs(integral-2672.448150653717)
    print("Integral=",integral, "Erro=", erro)
```

**Exemplo 2:** ([Fonte](https://integrada.minhabiblioteca.com.br/reader/books/9788580555691/pageid/29)) A segunda lei de Newton (a força $F$ é igual a massa $m$ vezes a aceleração $a$) pode ser usada para determinar a velocidade $v$ de um corpo em queda livre, perto da superfície da Terra

$$
F = ma \quad \rightarrow \quad a=\frac{F}{m} \quad \rightarrow \quad \frac{d v}{d t}=\frac{F}{m}
$$

Nesse caso, a força resultante é composta de duas forças opostas: a força gravitacional, para baixo, $F_D$ e a força da resistência do ar, para cima, $F_U$, ou seja:

$$
F=F_D+U_F \quad \text{onde} \quad F_D=m g \quad \text{e} \quad  F_U=-cv  
$$

onde $g$ é a constante gravitacional, ou a aceleração devida à gravidade (aproximadamente $9,81 \mathrm{~m} / \mathrm{s}^2$) e $c$ é uma constante de proporcionalidade chamada de *coeficiente de arrasto* (kg/s). 

Assim podemos escrever 

$$
\frac{d v}{d t}=\frac{m g-c v}{m} \quad \text{ou, simplesmente} \quad \frac{d v}{d t}=g-\frac{c}{m} v
$$

Esta equação é um modelo que relaciona a aceleração do objeto em queda às forças agindo nele.

Se o paraquedista estiver inicialmente em repouso ou seja $v=0$ em $t=0$, a função 

$$
v(t)=\frac{g m}{c}\left(1-e^{-(c / m) t}\right)
$$

satisfaz a equação do modelo, então, dizemos que $v(t)$ é uma solução do modelo. Vamos verificar essa afirmação no exemplo a seguir.
