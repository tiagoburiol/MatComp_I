```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl
plt.rcParams["font.family"] = "serif"
plt.rcParams["mathtext.fontset"] = "dejavuserif"
plt.rcParams["mathtext.default"]= "regular"
```

```python
x = np.linspace(-3, 3.3)
f = lambda x: 0.3*(x+1)*(x-2.1)*(1.1*x)-2
a = 1
b = 3
```

```python
# Ponto de interseção da reta com o eixo x (falsa posição)
xp =  b - f(b)*(a - b)/(f(a) - f(b))
xp
```

```python
fig, ax = plt.subplots()
ax.plot(x, f(x),'-' ,color='black')
ax.plot([2.705],f(2.705),'o',color='black')

ax.plot([2.2708], 0,'o',color='black')


# Linha da interpolação
ax.plot([a, b], [f(a),f(b)], 'k', linewidth=1)

# Área hachurada
ax.fill_between([a, b], [0, 0], [f(a), f(b)], color='powderblue')

ax.text(-0.15,-0.4, "O", style='italic')
ax.text(1.05,-0.4, "a", style='italic')
ax.text(2.705,f(2.705)-0.5, "$\\overline{x}$", style='italic')
ax.text(2.2708,-0.5, "$x_{f}$", style='italic')

ax.text(3,-0.4, "b", style='italic')

ax.text(-0.3,f(1)-0.15, "f(a)", style='italic')
ax.text(-0.3,f(3)-0.15, "f(b)", style='italic')


ax.vlines(x=[a,b], ymin=[0,0], ymax=[f(1),f(3)], 
          color='black', ls='--', lw=1)

ax.hlines(y=[f(a),f(b)],xmin=[0,0], xmax=[1,3], 
          color='black', ls='--', lw=1)


           
# Customizações para a figura final:
#____________________
# Limites em x e em y
ymin = -4; ymax = 5
xmin = -1; xmax = 3.8
plt.ylim(ymin, ymax) 
plt.xlim(xmin, xmax)
#________________________________________________
# Inclui setas nos eixos nas posições certas
ax.plot((xmax-0.07), (0), marker=">", color='black')
ax.text(xmax-0.1,-0.5, "x", style='italic')
ax.plot((0), (ymax-0.1), marker="^",color='black')
ax.text(-0.15, ymax-0.2, "y", style='italic')

# remove as bordas
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)

# desenha os eixos
ax.axhline(linewidth=1.0, color='black')
ax.axvline(linewidth=1.0, color='black')


plt.xticks([])
plt.yticks([])
plt.grid()
plt.savefig('falsa_posicao.png',dpi=300)
```
