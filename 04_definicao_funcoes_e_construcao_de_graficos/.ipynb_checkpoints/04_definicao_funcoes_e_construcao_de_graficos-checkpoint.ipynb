## 2. Funções e gráficos

Em matemática costumamos definir uma função $y=f(x)$ como uma regra que relaciona cada elemento $x$ de um conjunto $D_f$ (domínio de $f$), a um único elemento $y$ de um outro conjunto $CD_f$ (contradomínio). 

$$f:D_f \rightarrow CD_f$$

Essa relação geralmente é representada por uma equação matemática, mas também pode ser expressa por meio de palavras, por uma tabela ou um gráfico. Podemos dizer com isso que a função "transforma" $x$ em $y$, vamos imaginar que $x$ é a entrada e $y$ é a saída. O conjunto de todos elementos do contradomínio que estão relacionado a algum elemento do domínio chama-se imagem de $f$ e é representado por $Im_f$.

Em computação, uma função é uma sequência de instruções que realiza uma determinada tarefa. Basicamente consiste em um bloco de código que é executado quando a função é chamada. Geralmente uma função requer um ou mais argumentos de entrada e retorna uma saída. Assim, também podemos dizer que em computação uma função "transforma" algo ($x$) em uma outra coisa ($y$). 

Em Python, assim como em outras linguagens de programação, existem funções prontas e também módulos e bibliotecas de funções prontas. Temos especial interesse nas funções matemáticas que fazem parte da biblioteca `math`. Também utilizaremos muitas funções da bibliotema `numpy` para cálculo munérico e `matplotlib` para gráficos. Além das funções prontas, podemos definir nossas próprias funções usando `def`, como vamos ver no exemplo a seguir.

#### Exemplo 1

O aluguel de uma diária em uma casa de praia custa R$\$350,00$ mais R$\$50,00$ por pessoa e pode ser ocupada por no mínimo 2 pessoas e no máximo 7. Assim, o preço da diária $P$ é uma função do número $n$ de pessoas que irão ocupar a casa e pode ser expressa pela equação $P(n)=250+50n$ em que $n$ pertence ao conjunto ${2,3,4,5,6,7}$ e para cada possível valor de $n$ teremos um preço, tal que:

$P(2)=350+2 \times 50 = 450$

$P(3)=350+3 \times 50 = 500$

$P(4)=350+4 \times 50 = 550$

$P(5)=350+5 \times 50 = 600$

$P(6)=350+6 \times 50 = 650$

$P(7)=350+7 \times 50 = 700$

Nesse caso, tem-se $D_f=\{2,3,4,5,6,7\}$ e $Im_f=\{450,500,550,600,650,700\}$. O conjunto de pares ordenados $(n,P(n))$, $n \in D_f$ poderiam ser representados como pontos em um plano cartesiano, nesse caso teríamos a representação gráfica desta função.

Usando Python podemos facilmente obter esses resultados e representar esta função e seu gráfico da seguinte maneira:

```python
# Criamos a função
def P(n):
    return 350 + n*50
```

```python
# Calculamos os valores
Df = [2,3,4,5,6,7] # domínio
Imf = []           # lista vazia para a imagem

# para cada n calcula P(n) e adiciona 
for n in Df:
    Imf.append(P(n))
    print('P(',n,')=', Imf[-1])
```

É facil fazer o gráfico de pontos armazenados em uma lista. Para isso vamos importar o submódulo `pyplot` da bibliteca para gráficos `matplotlib`. Por simplicidade, daqui para frente vamos nos referir a esse submódulo como `plt` como é mostrado na linha abaixo.

```python
import matplotlib.pyplot as plt

```

Agora basta usarmos a função `plot`, passando os pontos (conjuntos domínio e imagem), para obter o gráfico. Se queremos apenas os pontos, basta incluir `"o"` e para desenhar as linhas de grade usamos `plt.grig()`. O resultado é mostrado abaixo. 

```python
# Plotamos os pontos
plt.plot(Df, Imf, 'o')
plt.grid() #mostra uma grade
```

#### Exemplo 2
Vamos começar plotando o gráfico de $f(x) = x^2-7x+3$, para isso precisamos definir a função e criar uma lista de pontos $(x, f(x))$ para plotar. Podemos fazer isso como é mostrado abaixo, vamos usar 21 pontos entre 10 e 10.

```python
xi = np.linspace(-10,10, 21)
f = lambda x: x**2-7*x+3
plt.plot(xi,f(xi), '.-')
plt.grid()
```

Observe que podemos estimar as raízes olhando para o gráfico e confirmar o resultado obtido no capítulo anterior. Abaixo mostramos como podemos visualizar os 5 primeiros pontos da tabela usada para plotar o gráfico.

```python
print('xi:', xi[0:5])
print('yi:', f(xi[0:5]))
```

#### Exemplo 2
Vamos agora plotar o gráfico de $f(d) = 1.047d^3-31.415d^2+2672.369$ modificando um pouco o código anterior.

```python
di = np.linspace(-20,40)
f = lambda d: 1.047*d**3-31.415*d**2+2672.369

plt.plot(di,f(di), '-k')
plt.grid()
```

#### Exemplo 3
Vamos agora modificar os códigos anteriores e plotar algunas funções legais. Você consegue identificar que funções são essas pelo código?

```python
xi = np.linspace(-10,10)
f0 = lambda x: np.exp(x)*np.sin(x)
f1 = lambda x: np.cos(x)*np.log(x+10.1)
f2 = lambda x: np.arctan(x)*np.sqrt(x+10)


fig, axs = plt.subplots(nrows=1, ncols=3, figsize=(12, 4))


axs[0].plot(xi,f0(xi), '-k')
axs[0].grid()
         
axs[1].plot(xi,f1(xi), '-k')
axs[1].grid()

axs[2].plot(xi,f2(xi), '-k')
axs[2].grid()
```

### Modelos matemáticos: algumas funções básicas

**Modelos lineares:** Quando $y$ é uma função linear de $x$, o gráfico é uma reta e pode ser descrito pela forma _inclinação-intersecção_,como

$$y = f(x) = mx + b$$

onde $m$ é o coeficiente angular da reta e $b$ é a intersecção com o eixo $y$. As funções lineares variam a uma taxa constante, o que significa que tomando dois pontos quaisquer da reta $(x_0, y_0)$ e $(x_1, y_1)$, com $x_1>x_0$ o valor de 

$$ \frac{\Delta y}{\Delta x} = \frac{y_1 - y_0}{x_1 - x_0} = m$$

é constante e ingual a $\tan(\alpha)$ onde $\alpha$ é o ângulo que a reta faz com um eixo horizontal. Vejamos um exemplo.
